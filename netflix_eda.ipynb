```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv("netflix_titles.csv.zip")

df.head()
df.info()
df.isnull().sum()
df.dropna(inplace=True)
df.drop_duplicates(inplace=True)
df.describe()
sns.countplot(x='type', data=df)
plt.title("Movies vs TV Shows")
plt.show()
top_country = df['country'].value_counts().head(10)
top_country.plot(kind='bar')
plt.title("Top 10 Countries")
plt.show()
df['date_added'] = pd.to_datetime(
    df['date_added'],
    errors='coerce'
)
df['year_added'] = df['date_added'].dt.year
df['year_added'].value_counts().sort_index().plot(kind='line')
plt.title("Content Added Over Years")
plt.xlabel("Year")
plt.ylabel("Count")
plt.show()
sns.countplot(y='rating',
              data=df,
              order=df['rating'].value_counts().index)
plt.title("Ratings Distribution")
plt.show()
df['listed_in'].value_counts().head(10).plot(kind='barh')
plt.title("Top Genres")
plt.show()
numeric_df = df[['release_year']]
sns.heatmap(numeric_df.corr(),
            annot=True)
plt.title("Correlation Heatmap")
plt.show()
```

# Final Insights

1. Netflix has a significantly higher number of Movies compared to TV Shows.

2. The United States contributes the largest share of content available on Netflix.

3. Netflix content additions increased rapidly after 2015, showing major platform growth.

4. TV-MA is the most common content rating, indicating a strong focus on mature audiences.

5. Drama and International Movies are among the most popular genres on the platform.
